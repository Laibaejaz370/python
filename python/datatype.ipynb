```python
# data type
# 1. string
name = "laiba"
print(name)  # laiba
```

```python
# numeric types
# int
age = 20
print(type(age))
print(age)
# float
height = 5.8
print(type(height))
print(height)
# complex
complex_num = 3 + 4j
print(type(complex_num))
print(complex_num)
```

```python
# sequence types
# list
fruits = ["apple", "banana", "cherry"]
print(type(fruits))
print(fruits)
shopping_list = ["milk", "bread", "eggs"]
shopping_list.append("butter")   # add new item
shopping_list.remove("bread")    # remove item
print(shopping_list)
# tuple
# Tuple – Car information
car = ("Toyota", "Corolla", 2022)
# Accessing tuple items
print("Car Brand:", car[0])
print("Car Model:", car[1])
print("Manufacture Year:", car[2])
# range
numbers = range(1, 10)
print(type(numbers))
print(list(numbers))
```

```python
# mapping types
# 01
dict = {"name": "Laiba", "age": 20, "city": "Gujranwala"}
print(type(dict))
print(dict)
# 02
# Dictionary – student info
student = {
    "name": "Ali",
    "age": 20,
    "grade": "A",
    "city": "Gujranwala"
}
# Accessing values
print("Name:", student["name"])
print("Age:", student["age"])
print("Grade:", student["grade"])
print("City:", student["city"])
```

```python
# set types
# set
# changeable
my_set = {1, 2, 3, 4, 5}
print(type(my_set))
print(my_set)
students = {"Ali", "Sara", "Ahmed"}
students.add("Laiba")   # add new student
print(students)
# frozenset
# not changeable
my_frozenset = frozenset([1, 2, 3, 4, 5])
print(type(my_frozenset))
print(my_frozenset)
team = frozenset(["Ali", "Sara", "Ahmed"])
print(team)
```

```python
# Boolean example
is_raining = True
has_umbrella = False

if is_raining and not has_umbrella:
    print("Take an umbrella!")
else:
    print("No need for an umbrella.")
```

```python
# bytes: Python me immutable sequence of binary data hoti hai, jo mostly files aur network data ke liye use hoti h
b = b"Hello"
print(b)
b = b"ABC"
print(b[0])  # ASCII value of 'A'
```

```python
# bytearray Python me mutable sequence of bytes hoti hai, matlab isko change kiya ja sakta hai.
ba = bytearray(b"Hello")
print(ba)
# modifying bytearray
ba = bytearray(b"Hello")
ba[0] = 74  # ASCII 'J'
print(ba)
```

```python
# memoryview Python me bytes ya bytearray ka view hota hai, jisse data copy kiye bina access kar sakte hain.
b = b"Hello"
mv = memoryview(b)
print(mv)
# memoryview Python me bytes ya bytearray ka view hota hai, jisse data copy kiye bina access kar sakte hain. --- IGNORE ---
# Access memoryview elements
b = b"ABC"
mv = memoryview(b)
print(mv[0])  # ASCII value of 'A'
```
